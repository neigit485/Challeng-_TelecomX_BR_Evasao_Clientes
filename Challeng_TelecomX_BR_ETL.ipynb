# 📊 Relatório Final:

## 1. Introdução
O meu objetivo nesse projeto foi pegar os dados brutos da empresa, arrumar a bagunça e tentar descobrir **por que** os clientes estão indo embora. A ideia é dar insights para a equipe de Data Science criar estratégias para segurar essa galera.

## 2. Limpeza e Tratamento dos Dados
Quando peguei os dados na API, eles estavam em formato JSON e bem aninhados (tinha dicionário dentro de dicionário). Tive que fazer uma faxina pesada:

1.  **Organização:** Usei o `json_normalize` para transformar o JSON numa tabela normal (DataFrame) que a gente consegue ler.
2.  **Tipos de Dados:** A coluna de *Cobrança Total* estava como texto (porque tinha espaços vazios). Transformei em número e preenchi os vazios com zero.
3.  **Padronização:** Traduzi as colunas do inglês para o português (ex: `tenure` virou `Meses_Contrato`) e mudei os valores (0 e 1) para "Ficou" e "Saiu" nos gráficos, pra ficar mais fácil de entender.
4.  **Criação de Variáveis:** Criei a coluna `Cobranca_Diaria` pra ter uma noção de quanto o cliente gasta por dia.

Agora o dataset está limpinho e pronto para análise!

```python
# Configurando o tamanho do painel
fig, axes = plt.subplots(1, 3, figsize=(20, 6))
sns.set(style="whitegrid")

# Gráfico 1: O vilão dos Contratos
sns.countplot(x='Contrato', hue='Churn', data=df_telecom, palette='viridis', ax=axes[0])
axes[0].set_title('Churn por Tipo de Contrato', fontsize=14)
axes[0].set_xlabel('Tipo de Contrato')
axes[0].set_ylabel('Quantidade de Clientes')
axes[0].legend(title='Status', labels=['Ficou', 'Saiu'])

# Gráfico 2: O problema do Cheque Eletrônico
sns.countplot(x='Metodo_Pagamento', hue='Churn', data=df_telecom, palette='magma', ax=axes[1])
axes[1].set_title('Churn por Método de Pagamento', fontsize=14)
axes[1].set_xlabel('Método de Pagamento')
axes[1].set_ylabel('')
axes[1].tick_params(axis='x', rotation=45)
axes[1].legend(title='Status', labels=['Ficou', 'Saiu'])

# Gráfico 3: Clientes novos saem mais? (Boxplot)
sns.boxplot(x='Churn', y='Meses_Contrato', data=df_telecom, palette='coolwarm', ax=axes[2])
axes[2].set_title('Distribuição de Tempo de Contrato', fontsize=14)
axes[2].set_xlabel('Status (0=Ficou, 1=Saiu)')
axes[2].set_ylabel('Meses de Contrato')
axes[2].set_xticklabels(['Ficou', 'Saiu'])

plt.tight_layout()
plt.show()
```

## 3. Principais Insights
Olhando para os gráficos acima e as estatísticas que rodamos, percebi coisas bem importantes:

* **A taxa de cancelamento é alta:** Cerca de **26,5%** dos clientes saíram. É mais de 1 em cada 4!
* **Contrato Mensal é perigoso:** A grande maioria de quem cancela tem contrato "Mensal". Quem assina contrato de 1 ou 2 anos quase não sai. Dá pra ver isso claramente no primeiro gráfico.
* **Atenção aos Novatos:** Quem sai, geralmente sai cedo. A mediana de tempo de casa de quem cancelou é de apenas **10 meses**. Quem passa desse período, tende a ficar.
* **Meio de Pagamento:** O "Cheque Eletrônico" tem uma taxa de cancelamento muito maior que cartão de crédito ou transferência. Pode ser que o sistema seja ruim ou o público que usa esse método seja menos fiel.
* **Preço:** Clientes com mensalidades mais caras tendem a sair mais.

## 4. Recomendações
Baseado no que vi, aqui vão minhas sugestões para a Telecom X:

1.  **Focar na Fidelidade:** Tentar migrar os clientes do plano "Mensal" para planos anuais. Talvez oferecer um desconto se a pessoa fechar por 1 ano?
2.  **Cuidar do "Onboarding":** Os primeiros 10 meses são críticos. A empresa deveria criar uma régua de relacionamento mais próxima com quem acabou de chegar para eles não desistirem no começo.
3.  **Investigar o Cheque Eletrônico:** Vale a pena a equipe de UX dar uma olhada se é difícil pagar com cheque eletrônico, ou se podemos incentivar o cadastro no cartão de crédito automático.
4.  **Benefícios para quem gasta muito:** Para evitar que os clientes que pagam caro saiam, podemos oferecer serviços extras (como mais internet ou streaming grátis) para aumentar o valor percebido.
