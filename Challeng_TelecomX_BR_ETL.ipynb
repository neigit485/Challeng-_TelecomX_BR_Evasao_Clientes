```python
import pandas as pd
import requests
import json
```

```python
url = "https://raw.githubusercontent.com/ingridcristh/challenge2-data-science/main/TelecomX_Data.json"
```

```python
response = requests.get(url)
data_json = response.json()
```

```python
df_telecom = pd.json_normalize(data_json)
```

```python
print("Dados carregados com sucesso!")
display(df_telecom.head())
```

```python
# Converter 'Total' para numérico, forçando erros a virarem NaN (vazio)
df_telecom['account.Charges.Total'] = pd.to_numeric(df_telecom['account.Charges.Total'], errors='coerce')
print(df_telecom['account.Charges.Total'].dtypes)
```

```python
colunas_texto = df_telecom.select_dtypes(include=['object']).columns

for col in colunas_texto:
    df_telecom[col] = df_telecom[col].str.strip()
```

```python
df_telecom['account.Charges.Total'] = pd.to_numeric(df_telecom['account.Charges.Total'], errors='coerce')
```

```python
df_telecom['account.Charges.Total'] = df_telecom['account.Charges.Total'].fillna(0.0)
```

```python
print(f"Linhas com Churn vazio antes da remoção: {df_telecom[df_telecom['Churn'] == ''].shape[0]}")
```

```python
df_telecom = df_telecom[df_telecom['Churn'] != '']
```

```python
df_telecom.reset_index(drop=True, inplace=True)
```
